# 本地 Notebook 环境检查

在 VS Code 中选择本项目的 `.venv/bin/python`，然后运行全部单元格。

本文件只验证 Python 内核和本地计算，不采集数据、不启动训练。

## 1. 确认 Python 环境

```python
import sys
import platform
from pathlib import Path
from importlib.metadata import version

expected = Path.cwd() / ".venv"
assert Path(sys.prefix).resolve() == expected.resolve(), f"请选择项目内核：{expected}/bin/python"
print("Python:", platform.python_version())
print("Interpreter:", sys.executable)
print("Machine:", platform.machine())
for package in ["ipykernel", "nbclient", "nbformat", "numpy"]:
    print(f"{package}: {version(package)}")
```

## 2. 验证计算与输出

```python
import numpy as np

values = np.array([1, 2, 3, 4])
assert values.sum() == 10
print("内核执行和 NumPy 计算正常。")
print("训练状态：未启动。数据集方案：待确定。")
```
